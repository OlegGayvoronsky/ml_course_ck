## ex. 1

$$  
y = x^Tx,  \quad x \in \mathbb{R}^N
$$

$$
\frac{dy}{dx} = 2x
$$

Объяснение:

Возьмем дифференциал

$$
dy = d(x^T x) = dx^T x + x^T dx
$$

Так как
$$
dx^T x = x^T dx
$$

Получаем

$$
dy = 2x^T dx
$$

В итоге, по определению производной

$$
\frac{dy}{dx} = 2x
$$

## ex. 2

$$ y = tr(AB) \quad A,B \in \mathbb{R}^{N \times N} $$

$$
\frac{dy}{dA} = B^T
$$

Объяснение:

Возьмем дифференциал

$$
dy = d(\operatorname{tr}(AB)) = d(A^T, B)
$$

При этом

$$
d(A^T, B) = d(B^T, A)
$$

Тогда получаем

$$
dy = (B^T, dA)
$$

В итоге, по определению градиента:

$$
\frac{dy}{dA} = B^T
$$

## Question 3

Which of the following statements about $k$-Nearest Neighbor ($k$-NN) are true in a classification setting, and for all $k$? Select all that apply.
1. The decision boundary (hyperplane between classes in feature space) of the k-NN classifier is linear.
2. The training error of a 1-NN will always be lower than that of 5-NN.
3. The test error of a 1-NN will always be lower than that of a 5-NN.
4. The time needed to classify a test example with the k-NN classifier grows with the size of the training set.
5. None of the above.

*Your Answer:* 2, 4


*Your Explanation:*
- В первом случае не верно, так как разделяющая граница не обязана быть линейной. Например, если облако точек одного класса расположено внутри другого;
- Во втором случае верно, так как на train части при k=1 ближайшей к точке будет она сама, поэтому все классы будут верными;
- В третьем случае не верно. Использование большего числа точек при принятии решения о классе тестовой точки в общем случае может как улучшить, так и ухудшить результат. Если полагаться только на ближайшего соседа, то это может привести к подгону модели под выбросы. С другой стороны, если использовать слишком много соседей, то в этом случае далекие точки могут влиять на решение. Поэтому выбор параметра k зависит от плотности распределения точек и количества выбросов;
- В четвертом случае верно, так как в таком случае при классификации тестовой точки прийдется считать большее число расстояний до точек выборки.
